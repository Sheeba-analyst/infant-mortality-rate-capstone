```python
#This removes common issues: & → and, brackets,extra spaces, "district",lowercase
import re

def clean_district(x):
    x = str(x).lower().strip()

    # --- symbol normalization ---
    x = x.replace("&", " and ")
    x = x.replace("-", " ")
    x = x.replace(".", "")
    x = x.replace(",", " ")

    # --- remove brackets ---
    x = re.sub(r"\(.*?\)", "", x)

    # --- remove common words ---
    x = re.sub(r"\b(district|dist|the)\b", "", x)

    # --- remove admin suffixes (very important) ---
    x = re.sub(r"\b(urban|rural|suburban)\b", "", x)

    # --- normalize known patterns ---
    x = re.sub(r"\bnagar\b", " nagar", x)   # fix joined words
    x = re.sub(r"\bgarh\b", " garh", x)

    # --- fix numbers in words ---
    x = x.replace("twenty four", "24")
    x = x.replace("twenty four", "24")

    # --- remove extra words like 'and' (after standardization) ---
    x = re.sub(r"\band\b", " ", x)

    # --- normalize spaces ---
    x = re.sub(r"\s+", " ", x)

    return x.strip()
for df in [df1, df2_agg, df3_latest]:
    df['srcDistrictName'] = df['srcDistrictName'].apply(clean_district)
    df['srcStateName'] = df['srcStateName'].str.lower().str.strip()
```

## RQ3: Predicting IMR using ML Models

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_final, y, test_size=0.2, random_state=42
)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

rf = RandomForestRegressor(random_state=42)
rf.fit(X_train, y_train)

pred_rf = rf.predict(X_test)

rf_mae = mean_absolute_error(y_test, pred_rf)
rf_rmse = np.sqrt(mean_squared_error(y_test, pred_rf))
rf_r2 = r2_score(y_test, pred_rf)

print("RF → MAE:", rf_mae)
print("RF → RMSE:", rf_rmse)
print("RF → R2:", rf_r2)
```

```python
from sklearn.ensemble import GradientBoostingRegressor

gb = GradientBoostingRegressor()
gb.fit(X_train, y_train)

pred_gb = gb.predict(X_test)

gb_mae = mean_absolute_error(y_test, pred_gb)
gb_rmse = np.sqrt(mean_squared_error(y_test, pred_gb))
gb_r2 = r2_score(y_test, pred_gb)

print("GB → R2:", gb_r2)
```

```python
from xgboost import XGBRegressor

xgb = XGBRegressor()
xgb.fit(X_train, y_train)

pred_xgb = xgb.predict(X_test)

xgb_mae = mean_absolute_error(y_test, pred_xgb)
xgb_rmse = np.sqrt(mean_squared_error(y_test, pred_xgb))
xgb_r2 = r2_score(y_test, pred_xgb)

print("XGB → R2:", xgb_r2)
```

```python
import pandas as pd

results = pd.DataFrame({
    "Model": ["Random Forest", "Gradient Boosting", "XGBoost"],
    "R2": [rf_r2, gb_r2, xgb_r2],
    "RMSE": [rf_rmse, gb_rmse, xgb_rmse],
    "MAE": [rf_mae, gb_mae, xgb_mae]
})

print(results.sort_values(by="R2", ascending=False))
```

```python
def predict_and_explain(district_name):

    district_name = district_name.lower().strip()

    mask = final_df['srcDistrictName'] == district_name

    if mask.sum() == 0:
        return "District not found"

    row = X_final[mask].iloc[0]

    # Prediction
    pred = float(rf.predict([row])[0])   # convert to normal float

    # Explanation
    explanation = []
    for feature in top_features:
        explanation.append(f"{feature}: {round(row[feature],2)}")

    return {
        "District": district_name,
        "Predicted IMR": round(pred, 2),
        "Explanation": explanation
    }
```
