````python
#This removes common issues: & → and, brackets,extra spaces, "district",lowercase
import re

def clean_district(x):
    x = str(x).lower().strip()

    # --- symbol normalization ---
    x = x.replace("&", " and ")
    x = x.replace("-", " ")
    x = x.replace(".", "")
    x = x.replace(",", " ")

    # --- remove brackets ---
    x = re.sub(r"\(.*?\)", "", x)

    # --- remove common words ---
    x = re.sub(r"\b(district|dist|the)\b", "", x)

    # --- remove admin suffixes (very important) ---
    x = re.sub(r"\b(urban|rural|suburban)\b", "", x)

    # --- normalize known patterns ---
    x = re.sub(r"\bnagar\b", " nagar", x)   # fix joined words
    x = re.sub(r"\bgarh\b", " garh", x)

    # --- fix numbers in words ---
    x = x.replace("twenty four", "24")
    x = x.replace("twenty four", "24")

    # --- remove extra words like 'and' (after standardization) ---
    x = re.sub(r"\band\b", " ", x)

    # --- normalize spaces ---
    x = re.sub(r"\s+", " ", x)

    return x.strip()
for df in [df1, df2_agg, df3_latest]:
    df['srcDistrictName'] = df['srcDistrictName'].apply(clean_district)
    df['srcStateName'] = df['srcStateName'].str.lower().str.strip()
````

Outliers in IMR were identified using the IQR method. A small number of districts exhibited unusually high IMR values, likely due to low reported live births or reporting inconsistencies. These observations were retained for analysis as they reflect real-world disparities, but sensitivity checks were performed to ensure model robustness.

Outlier treatment was applied only to the target variable (IMR) to reduce the influence of extreme values on model performance. No outlier removal or transformation was performed on the predictor variables, as district-level indicators naturally exhibit high variability due to differences in population size, geographic conditions, and socio-economic characteristics. Treating such variations as outliers could remove meaningful real-world heterogeneity and lead to loss of important information. Therefore, all predictor variables were retained in their original form to preserve the true distribution of district-level differences.

### Global Moran’s I

````python
import libpysal
from esda.moran import Moran

mor_df = map_df.dropna(subset=['IMR_capped']).copy()

w = libpysal.weights.KNN.from_dataframe(mor_df, k=5)
w.transform = 'r'

moran = Moran(mor_df['IMR_capped'], w)

print("Moran's I =", moran.I)
print("p-value =", moran.p_sim)
````

Spatial autocorrelation of infant mortality rates was assessed using Moran’s I statistic. The analysis resulted in a Moran’s I value of 0.512 and a p-value of 0.001.

The positive Moran’s I value indicates a strong positive spatial autocorrelation, suggesting that districts with similar infant mortality rates are geographically clustered. In particular, districts with high infant mortality rates tend to be located near other high-IMR districts, while districts with lower IMR are similarly clustered together.

The statistically significant p-value confirms that this spatial clustering is unlikely to have occurred by chance. These findings highlight the presence of regional spatial patterns in infant mortality and suggest that geographic location plays an important role in influencing health outcomes.

This spatial dependence reinforces the need for geographically targeted public health interventions and supports the use of spatial analysis techniques in understanding infant mortality patterns.

Inference:

The Global Moran’s I statistic for Infant Mortality Rate (IMR) was 0.512 with a statistically significant p-value of 0.001, indicating strong positive spatial autocorrelation across Indian states. This suggests that states with similar IMR levels tend to be geographically clustered rather than randomly distributed. The result confirms the presence of significant spatial dependence in IMR patterns across the study region.

Also:

The positive Moran’s I value demonstrates that neighboring states exhibit comparable infant mortality characteristics, highlighting regional concentration effects and potential spatial spillover influences in healthcare and socio-economic conditions.

Interpretation scale commonly used:

Moran’s I	Interpretation
~0	Random spatial pattern
0.1 – 0.3	Weak spatial autocorrelation
0.3 – 0.5	Moderate spatial autocorrelation
> 0.5	Strong spatial autocorrelation

So the value (0.512) indicates strong spatial clustering.

````python
# =========================================================
# OPTION 1 — SHAP-BASED EXPLAINABILITY (Recommended)
# =========================================================

import shap
import pandas as pd
import numpy as np

# Use trained model (rf / gb / xgb)
explainer = shap.TreeExplainer(gb)

# SHAP values
shap_values = explainer.shap_values(X_final)

# ---------------------------------------------------------
# Predict + Explain Function
# ---------------------------------------------------------

def predict_and_explain(district_name):

    district_name = district_name.lower().strip()

    mask = final_df['srcDistrictName'].str.lower() == district_name

    if mask.sum() == 0:
        return "District not found"

    # district row
    row = X_final[mask].iloc[0]

    # prediction
    pred = float(gb.predict([row])[0])

    # SHAP values for district
    row_index = X_final[mask].index[0]

    district_shap = pd.Series(
        shap_values[row_index],
        index=X_final.columns
    )

    # top contributing features
    top_features = district_shap.abs() \
                                 .sort_values(ascending=False) \
                                 .head(5)

    explanation = []

    for feature in top_features.index:

        contribution = district_shap[feature]
        value = row[feature]

        direction = "increased" if contribution > 0 else "reduced"

        explanation.append(
            f"{feature} = {round(value,2)} "
            f"({direction} predicted IMR)"
        )

    return {
        "District": district_name.title(),
        "Predicted IMR": round(pred, 2),
        "Top Factors": explanation
    }

# Example
predict_and_explain("gaya")
````

````python
# =====================================================
# OPTION 6 — LIGHTWEIGHT RAG
# =====================================================

def rag_explanation():

    district_name = input(
        "Enter district name: "
    ).lower().strip()

    mask = (
        final_df['srcDistrictName']
        .str.lower()
        == district_name
    )

    if mask.sum() == 0:

        print("District not found")
        return

    row = X_final[mask].iloc[0]

    row_index = X_final[mask].index[0]

    pred = float(
        gb.predict([row])[0]
    )

    # -----------------------------------
    # SHAP
    # -----------------------------------

    district_shap = pd.DataFrame({

        "Feature": X_final.columns,

        "SHAP_Value":
        shap_values[row_index],

        "Feature_Value":
        row.values

    })

    district_shap["Abs_SHAP"] = (
        district_shap["SHAP_Value"]
        .abs()
    )

    district_shap = district_shap.sort_values(
        by="Abs_SHAP",
        ascending=False
    )

    top5 = district_shap.head(5)

    # -----------------------------------
    # RETRIEVAL
    # -----------------------------------

    top_features = (
        top5["Feature"]
        .tolist()
    )

    retrieved_context = (
        retrieve_context(
            top_features
        )
    )

    print("\nRetrieved Context:\n")
    print(retrieved_context)

    # -----------------------------------
    # PROMPT
    # -----------------------------------

    prompt = f"""
District: {district_name.title()}

Predicted IMR:
{round(pred,2)}

Top SHAP Features:

{top5[['Feature','SHAP_Value']]}

Retrieved Healthcare Knowledge:

{retrieved_context}

Explain:

1. Why the district has this predicted IMR

2. Which factors appear protective

3. Which factors appear associated
with higher risk

4. Do not imply causation

5. Use simple public-health language
"""

    response = client.chat.completions.create(

        model="gpt-4.1-mini",

        messages=[
            {
                "role":"user",
                "content":prompt
            }
        ]
    )

    print("\n======================")
    print("District:", district_name.title())
    print("Predicted IMR:", round(pred,2))
    print("======================\n")

    print(
        response
        .choices[0]
        .message
        .content
    )
````
